```python
import os
```

```python
import cv2
```

```python
import numpy as np
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
```

```python
from sklearn.model_selection import train_test_split
```

```python
from tensorflow.keras.callbacks import ReduceLROnPlateau
```

```python
# Constants
IMG_HEIGHT, IMG_WIDTH = 128, 128
X, y = [], []
```

```python
# Assign labels
label_map = {}  # Map (user_id, type) to a label
class_counter = 0
```

```python
# Process dataset
for user_id in range(1, 31): 
    genuine_path = f'signatures/user_{user_id}/genuine'
    forgery_path = f'signatures/user_{user_id}/forgery'

    # Assign labels for genuine and forgery
    label_map[(user_id, 'genuine')] = class_counter
    class_counter += 1
    label_map[(user_id, 'forgery')] = class_counter
    class_counter += 1

    # Process genuine signatures
    if os.path.exists(genuine_path):
        for filename in os.listdir(genuine_path):
            img_path = os.path.join(genuine_path, filename)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            img = cv2.resize(img, (IMG_HEIGHT, IMG_WIDTH))
            X.append(img)
            y.append(label_map[(user_id, 'genuine')])

    # Process forgery signatures
    if os.path.exists(forgery_path):
        for filename in os.listdir(forgery_path):
            img_path = os.path.join(forgery_path, filename)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            img = cv2.resize(img, (IMG_HEIGHT, IMG_WIDTH))
            X.append(img)
            y.append(label_map[(user_id, 'forgery')])

unknown_user_label = class_counter  
class_counter += 1


unknown_path = 'signatures/unknown'
if os.path.exists(unknown_path):
    for filename in os.listdir(unknown_path):
        img_path = os.path.join(unknown_path, filename)
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        img = cv2.resize(img, (IMG_HEIGHT, IMG_WIDTH))
        X.append(img)
        y.append(unknown_user_label)
```

```python
# import matplotlib.pyplot as plt
# for i in range(1,5):
#     plt.imshow(i.numpy().astype("unit8"))
#     plt.axis("off")

# # plt.figure(figsize=(10, 10))
# # for image_batch, label_batch in dataset.take(1):
# #     for i in range(1, 20):
# #         ax = plt.subplot(5, 4, i)
# #         plt.imshow(image_batch[i].numpy().astype("uint8"))
# #         plt.title(class_names[labels_batch[i]])
        
```

```python
# Convert to numpy arrays
X = np.array(X) / 255.0  # Normalize
X = X.reshape(-1, IMG_HEIGHT, IMG_WIDTH, 1)  # Reshape
y = np.array(y)
```

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(IMG_HEIGHT, IMG_WIDTH, 1)),
    MaxPooling2D(2, 2),
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    MaxPooling2D(2, 2),
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    MaxPooling2D(2, 2),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(class_counter + 1, activation='softmax')  # Total classes
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
# Train the model
lr_scheduler = ReduceLROnPlateau(monitor='val_loss', patience=2, factor=0.5, min_lr=1e-6)
# model.fit(X_train, y_train, batch_size=16, epochs=20, validation_data=(X_test, y_test))
```

```python
import matplotlib.pyplot as plt

# Train the model and capture the training history
history = model.fit(X_train, y_train, batch_size=16, epochs=20, validation_data=(X_test, y_test), callbacks=[lr_scheduler])

# Plot the training and validation accuracy
plt.figure(figsize=(12, 6))

# Accuracy plot
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid()

# Loss plot
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
# Prediction function
def predict_signature(test_signature_path):
    img = cv2.imread(test_signature_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.resize(img, (IMG_HEIGHT, IMG_WIDTH))
    img = img / 255.0
    img = img.reshape(1, IMG_HEIGHT, IMG_WIDTH, 1)

    prediction = model.predict(img)
    predicted_class = np.argmax(prediction)

    # Check if the predicted class is an unknown user
    if predicted_class == unknown_user_label:
        return "Unknown User"
    else:
        # Find the corresponding user and type
        for (user_id, type_), label in label_map.items():
            if label == predicted_class:
                return f"User {user_id} - {type_.capitalize()}"

# Example usage
test_signature_path = 'user_10_forgery_3.png'
result = predict_signature(test_signature_path)
print(f"The signature is: {result}")

# Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix

# Ensure target names match actual classes
target_names = [f"User {user_id} - {type_}" for (user_id, type_) in label_map.keys()]
target_names.append("Unknown User")  # Add the unknown user label explicitly

# Predict on the test data
y_pred = model.predict(X_test)
y_pred = np.argmax(y_pred, axis=1)  # Convert probabilities to class labels

# Generate confusion matrix
cm = confusion_matrix(y_test, y_pred)

# For multi-class classification
TP = np.sum(np.diag(cm)) 
Total = np.sum(cm)  # Total samples
FN_FP = np.sum(cm) - TP  


TN = Total - FN_FP - TP
FP_FN = FN_FP

# Print results
print("Confusion Matrix:\n", cm)
print("True Positives (TP):", TP)
print("True Negatives (TN):", TN)
print("False Positives (FP):", FP_FN)
print("False Negatives (FN):", FP_FN)
print("Total  :", Total)


# Optionally, print classification report
from sklearn.metrics import classification_report
report = classification_report(y_test, y_pred, target_names=target_names)
print("Classification Report:\n", report)
```
